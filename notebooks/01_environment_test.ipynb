```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy

# Generate sample customer demand
demand = np.random.poisson(lam=120, size=10)

print("Sample Customer Demand:")
print(demand)

# Convert into a DataFrame
df = pd.DataFrame(demand, columns=["Customer_Demand"])

print("\nDemand Dataset:")
display(df)

# Visualize demand
plt.hist(demand)
plt.title("Sample Customer Demand")
plt.xlabel("Demand")
plt.ylabel("Frequency")
plt.show()
```
