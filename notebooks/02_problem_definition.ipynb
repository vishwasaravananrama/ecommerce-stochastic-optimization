```python
## 2. Real-World Context

An e-commerce seller orders headphones from a supplier **before** a sales period
(for example, a festival sale or a month). Actual demand is unknown at that time.

| If the seller orders **too many** | If the seller orders **too few** |
|---|---|
| Unsold stock remains | Some customers cannot be served |
| Holding cost is paid | Shortage penalty is incurred (service / reputation damage) |
| Money is tied up in excess stock | Potential sales are missed |

Ordering too much or too little both cost money, so the best choice is a balance.
```
